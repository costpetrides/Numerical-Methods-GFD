## Άσκηση — Μονοδιάστατη Εξίσωση Θερμότητας

Δίνεται η μονοδιάστατη εξίσωση θερμότητας:

$$
\frac{\partial u}{\partial t}
=
\frac{\partial^2 u}{\partial x^2},
\qquad
0<x<1,\quad 0<t\leq 0.1
$$

με αρχική συνθήκη:

$$
u(x,0)=
\begin{cases}
2x, & 0\leq x\leq 0.5,\\[4pt]
2(1-x), & 0.5<x\leq 1
\end{cases}
$$

και συνοριακές συνθήκες:

$$
u(0,t)=0,
\qquad
u(1,t)=0.
$$

Να επιλυθεί αριθμητικά η εξίσωση με τη μέθοδο **FTCS (Forward-Time Central-Space)**, χρησιμοποιώντας χωρικό και χρονικό βήμα:

   $$
   h=0.1,\qquad k=0.001.
   $$

Να υπολογιστεί η αριθμητική λύση τη χρονική στιγμή $t=0.05$ και να παρουσιαστεί η αριθμητική λύση $u(x,t)$ σε τρισδιάστατο διάγραμμα επιφάνειας.


```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
L = 1
T = 0.1
h = 0.1
k = 0.001

# Computational grid
x = np.arange(0, L+h, h)
t = np.arange(0, T+k, k)

Nx = len(x)
Nt = len(t)

# Stability condition
r = k/h**2
assert r <= 0.5, "FTCS stability condition violated."

# Solution matrix (time x space)
u = np.zeros((Nt, Nx))

# Initial condition (two intervals)
u[0, x <= 0.5] = 2*x[x <= 0.5]
u[0, x > 0.5] = 2*(1-x[x > 0.5])

# Boundary conditions
u[:, 0] = 0
u[:, -1] = 0

# FTCS method
i = np.arange(1, Nx-1)

for j in range(Nt-1):
    u[j+1, i] = u[j, i] + r*(
        u[j, i-1] - 2*u[j, i] + u[j, i+1]
    )

# Solution matrix
print(u)

# Solution at t = 0.05
j = round(0.05/k)

print(f"Solution at t = {t[j]:.3f}")
print(u[j, :])

# Visualization
X, Time = np.meshgrid(x, t)

fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")

ax.plot_surface(X, Time, u)

ax.set_xlabel("x")
ax.set_ylabel("t")
ax.set_zlabel("u(x,t)")
ax.set_title("1D Heat Equation - FTCS")

plt.show()
```
