```python
import numpy as np
import matplotlib.pyplot as plt

# Grid
L = 2*np.pi # L=b-a [a,b]
N = 100

x = np.linspace(0, L, N+1)
h = x[1] - x[0]

# Function
u = np.sin(x)

# Finite Differences
du_forward = (u[2:] - u[1:-1]) / h
du_backward = (u[1:-1] - u[:-2]) / h
du_centered = (u[2:] - u[:-2]) / (2*h)

# Plot
plt.figure(figsize=(8, 4))

plt.plot(x[1:-1], du_forward, label="Forward")
plt.plot(x[1:-1], du_backward, label="Backward")
plt.plot(x[1:-1], du_centered, label="Centered")

plt.xlabel("x")
plt.ylabel("du/dx")
plt.legend()
plt.grid(True)
plt.show()
```
