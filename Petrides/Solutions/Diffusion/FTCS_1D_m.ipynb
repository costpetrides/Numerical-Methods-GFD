##  Άσκηση 1.

```matlab
clc; clear; close all;

L = 1;
T = 0.1;
h = 0.1;
k = 0.002;

x = (0:h:L)';
t = 0:k:T;

Nx = numel(x);
Nt = numel(t);

r = k/h^2;

assert(r <= 0.5, 'FTCS is unstable');

u = zeros(Nx,Nt);

% Initial condition
u(:,1) = 20 + 40*x + 30*sin(pi*x);

% Dirichlet boundary conditions
u(1,:) = 20;
u(end,:) = 60;

% FTCS method
i = 2:Nx-1;

for j = 1:Nt-1
    u(i,j+1) = u(i,j) + r*( ...
        u(i-1,j) - 2*u(i,j) + u(i+1,j));
end

% Contour
figure();
contourf(x,t,u',25,'LineColor','none');
colormap(turbo);
colorbar;
xlabel('Position x');
ylabel('Time t');
title('FTCS - Contour');
axis tight;
```

```matlab
clc; clear; close all;

L = 1;
T = 0.1;
h = 0.05;
k = 0.001;

x = (0:h:L)';
t = 0:k:T;

Nx = numel(x);
Nt = numel(t);

r = k/h^2;

assert(r <= 0.5, 'FTCS is unstable');

u = zeros(Nx,Nt);

% Initial condition
u(:,1) = 20 + 40*x + 30*sin(2*pi*x) ...
         + 20*x.*(1-x);

% Dirichlet boundary conditions
u(1,:) = 20 + 10*sin(20*pi*t);
u(end,:) = 60 + 100*t;

% FTCS method
i = 2:Nx-1;

for j = 1:Nt-1
    u(i,j+1) = u(i,j) + r*( ...
        u(i-1,j) - 2*u(i,j) + u(i+1,j));
end

% Value at x = 0.35, t = 0.05
ix = round(0.35/h) + 1;
jt = round(0.05/k) + 1;

fprintf('u(0.35,0.05) = %.6f\n',u(ix,jt));
```

## Άσκηση 3.

```matlab
clc; clear; close all;

L = 2;
T = 0.2;
h = 0.05;
k = 0.001;

x = (0:h:L)';
t = 0:k:T;

Nx = numel(x);
Nt = numel(t);

r = k/h^2;

assert(r <= 0.5, 'FTCS is unstable');

u = zeros(Nx,Nt);

% Initial condition
u(x <= 0.5,1) = 20 + 80*x(x <= 0.5).^2;

u(x > 0.5 & x <= 1.5,1) = ...
    60 + 40*sin(pi*(x(x > 0.5 & x <= 1.5)-0.5));

u(x > 1.5,1) = ...
    100 - 80*(x(x > 1.5)-1.5).^2;

% Dirichlet boundary conditions
u(1,:) = 20 + 15*sin(10*pi*t);
u(end,:) = 80 + 20*t;

% FTCS method
i = 2:Nx-1;

for j = 1:Nt-1
    u(i,j+1) = u(i,j) + r*( ...
        u(i-1,j) - 2*u(i,j) + u(i+1,j));
end

% Contour
figure();
contourf(x,t,u',25,'LineColor','none');
colormap(turbo);
colorbar;
xlabel('Position x');
ylabel('Time t');
title('FTCS - Contour');
axis tight;
```

```matlab
% Maximum temperature
umax = max(u(:));
[ix,jt] = find(u == umax,1);

fprintf('Max = %.2f at x = %.2f, t = %.3f\n', ...
    umax,x(ix),t(jt));
```


## Άσκηση 4.

```matlab
clc; clear; close all;

L = 2;
T = 0.3;
h = 0.05;
k = 0.001;

x = (0:h:L)';
t = 0:k:T;

Nx = numel(x);
Nt = numel(t);

r = k/h^2;

assert(r <= 0.5, 'FTCS is unstable');

u = zeros(Nx,Nt);

% Initial condition
u(:,1) = 50;

% Left Dirichlet boundary
a = t <= 0.1;
b = t > 0.1;

u(1,a) = 50 + 20*sin(5*pi*t(a));
u(1,b) = 70 - 100*(t(b)-0.1);

% Right Dirichlet boundary
a = t <= 0.15;
b = t > 0.15;

u(end,a) = 50 + 100*t(a);
u(end,b) = 65 - 200*(t(b)-0.15).^2;

% FTCS method
i = 2:Nx-1;

for j = 1:Nt-1
    u(i,j+1) = u(i,j) + r*( ...
        u(i-1,j) - 2*u(i,j) + u(i+1,j));
end

% Maximum temperature
umax = max(u(:));
[ix,jt] = find(u == umax,1);

fprintf('Max = %.2f at x = %.2f, t = %.3f\n', ...
    umax,x(ix),t(jt));
```

## Άσκηση 5.

```matlab
clc; clear; close all;

L = 1;
T = 0.2;
h = 0.05;
k = 0.001;

x = (0:h:L)';
t = 0:k:T;

Nx = numel(x);
Nt = numel(t);

r = k/h^2;

assert(r <= 0.5, 'FTCS is unstable');

u = zeros(Nx,Nt);

% Initial condition
u(:,1) = 20 + 40*x;

% Dirichlet boundary conditions
u(1,:) = 20;
u(end,:) = 60;

% FTCS method with source term
i = 2:Nx-1;

for j = 1:Nt-1
    S = 30*exp(-5*t(j))*sin(pi*x(i));

    u(i,j+1) = u(i,j) + r*( ...
        u(i-1,j) - 2*u(i,j) + u(i+1,j)) ...
        + k*S;
end

% Contour
figure();
contourf(x,t,u',25,'LineColor','none');
colormap(turbo);
colorbar;
xlabel('Position x');
ylabel('Time t');
title('FTCS - Diffusion with Source Term');
axis tight;
```

## Άσκηση 6.

Ο όρος πηγής υπολογίζεται από τη γνωστή τιμή $u_i^j$


στο προηγούμενο χρονικό επίπεδο. Έτσι, η μέθοδος


παραμένει ρητή (Explicit FTCS).


$$u_i^{j+1} =u_i^j +r(u_{i-1}^j -2u_i^j +u_{i+1}^j )+3ku_i^j$$

```matlab
clc; clear; close all;

L = 1;
T = 0.2;
h = 0.05;
k = 0.001;

x = (0:h:L)';
t = 0:k:T;

Nx = numel(x);
Nt = numel(t);

r = k/h^2;

assert(r <= 0.5, 'FTCS is unstable');

u = zeros(Nx,Nt);

% Initial condition
u(:,1) = 20 + 40*x;

% Dirichlet boundary conditions
u(1,:) = 20;
u(end,:) = 60;

% FTCS method with source term
i = 2:Nx-1;

for j = 1:Nt-1
    u(i,j+1) = u(i,j) + r*( ...
        u(i-1,j) - 2*u(i,j) + u(i+1,j)) ...
        + 3*k*u(i,j);
end

% Contour
figure();
contourf(x,t,u',25,'LineColor','none');
colormap(turbo);
colorbar;
xlabel('Position x');
ylabel('Time t');
title('FTCS - Diffusion with Source Term');
axis tight;
```

## Άσκηση 7.

Εφαρμόζουμε τη μέθοδο FTCS:


 $u_i^{j+1} =u_i^j +r(u_{i-1}^j -2u_i^j +u_{i+1}^j )$      όπου $r=\frac{k}{h^2 }$.


$$u_1^{j+1} =u_1^j +r(u_0^j -2u_1^j +u_2^j )$$

Δεν γνωρίζουμε το $u_o$ επομένως πρέπει να το υπολογίσουμε μέσο των συνοριακών!


Εφαρμόζουμε Centered Difference στην οριακή συνθήκη:


$$\frac{u_2^j -u_0^j }{2h}=u_1^j$$

Λύνουμε ως προς το εικονικό σημείο:


$$u_0^j =u_2^j -2hu_1^j$$

Αντικαθιστούμε στον τύπο FTCS:


$$u_1^{j+1} =u_1^j +r(u_2^j -2hu_1^j -2u_1^j +u_2^j )$$

Ομαδοποιώντας τους όρους, προκύπτει:


$$u_1^{j+1} =u_1^j +2r[u_2^j -(1+h)u_1^j ]$$

Εφαρμόζουμε Centered Difference στην οριακή συνθήκη:


$$\frac{u_{N_x +1}^j -u_{N_x -1}^j }{2h}=-u_{N_x }^j$$

Λύνουμε ως προς το εικονικό σημείο:


$$u_{N_x +1}^j =u_{N_x -1}^j -2hu_{N_x }^j$$

Αντικαθιστούμε στον τύπο FTCS:


$$u_{N_x }^{j+1} =u_{N_x }^j +r(u_{N_x -1}^j -2u_{N_x }^j +u_{N_x -1}^j -2hu_{N_x }^j )$$

Ομαδοποιώντας τους όρους, προκύπτει:


$$u_{N_x }^{j+1} =u_{N_x }^j +2r[u_{N_x -1}^j -(1+h)u_{N_x }^j ]$$

```matlab
clc; clear; close all;

L = 1;
T = 0.1;
h = 0.1;
k = 0.001;

x = (0:h:L)';
t = 0:k:T;

Nx = numel(x);
Nt = numel(t);

r = k/h^2;

assert(r <= 1/(2*(1+h)), 'FTCS stability condition violated');

u = zeros(Nx,Nt);

% Initial condition
u(:,1) = 1;

% FTCS method
i = 2:Nx-1;

for j = 1:Nt-1

    % Interior points
    u(i,j+1) = u(i,j) + r*( ...
        u(i-1,j) - 2*u(i,j) + u(i+1,j));

    % Left Robin boundary
    u(1,j+1) = u(1,j) + 2*r*( ...
        u(2,j) - (1+h)*u(1,j));

    % Right Robin boundary
    u(end,j+1) = u(end,j) + 2*r*( ...
        u(end-1,j) - (1+h)*u(end,j));

end

% Contour
figure();
contourf(x,t,u',25,'LineColor','none');
colormap(turbo);
colorbar;
xlabel('Position x');
ylabel('Time t');
title('FTCS - Robin Boundary Conditions');
axis tight;
```

## Άσκηση 8.

Στα εσωτερικά σημεία εφαρμόζουμε τη μέθοδο FTCS:


$$u_i^{j+1} =u_i^j +r(u_{i-1}^j -2u_i^j +u_{i+1}^j )$$

όπου $r=\frac{k}{h^2 }$


Εφαρμόζουμε Centered Difference στην οριακή συνθήκη:


$$\frac{u_2^j -u_0^j }{2h}=10+5\cos \left(\frac{\pi t_j }{T}\right)$$

Λύνοντας ως προς το εικονικό σημείο:


$$u_0^j =u_2^j -2h\left(10+5\cos \frac{\pi t_j }{T}\right)$$

Αντικαθιστούμε στον τύπο FTCS:


$$u_1^{j+1} =u_1^j +r(u_0^j -2u_1^j +u_2^j )$$

$$=u_1^j +r\left\lbrack 2u_2^j -2u_1^j -2h\left(10+5\cos \frac{\pi t_j }{T}\right)\right\rbrack$$

Ομαδοποιώντας τους όρους, προκύπτει:


$$u_1^{j+1} =u_1^j +2r\left\lbrack u_2^j -u_1^j -h\left(10+5\cos \frac{\pi t_j }{T}\right)\right\rbrack$$

Εφαρμόζουμε Centered Difference στην οριακή συνθήκη:


$$\frac{u_{N_x +1}^j -u_{N_x -1}^j }{2h}=10-5\sin \left(\frac{\pi t_j }{T}\right)$$

Λύνοντας ως προς το εικονικό σημείο:


$$u_{N_x +1}^j =u_{N_x -1}^j +2h\left(10-5\sin \frac{\pi t_j }{T}\right)$$

Αντικαθιστούμε στον τύπο FTCS:


$$u_{N_x }^{j+1} =u_{N_x }^j +r(u_{N_x -1}^j -2u_{N_x }^j +u_{N_x +1}^j )$$

$$=u_{N_x }^j +r\left\lbrack 2u_{N_x -1}^j -2u_{N_x }^j +2h\left(10-5\sin \frac{\pi t_j }{T}\right)\right\rbrack$$

Ομαδοποιώντας τους όρους, προκύπτει:


$$u_{N_x }^{j+1} =u_{N_x }^j +2r\left\lbrack u_{N_x -1}^j -u_{N_x }^j +h\left(10-5\sin \frac{\pi t_j }{T}\right)\right\rbrack$$

```matlab
clc; clear; close all;

L = 1;
T = 0.2;
h = 0.05;
k = 0.001;

x = (0:h:L)';
t = 0:k:T;

Nx = numel(x);
Nt = numel(t);

r = k/h^2;

assert(r <= 0.5, 'FTCS is unstable');

u = zeros(Nx,Nt);

% Initial condition
u(:,1) = 20 + 10*x + 5*sin(pi*x);

% FTCS method
i = 2:Nx-1;

for j = 1:Nt-1

    % Interior points
    u(i,j+1) = u(i,j) + r*( ...
        u(i-1,j) - 2*u(i,j) + u(i+1,j));

    % Left Neumann boundary
    u(1,j+1) = u(1,j) + 2*r*( ...
        u(2,j) - u(1,j) ...
        - h*(10 + 5*cos(pi*t(j)/T)));

    % Right Neumann boundary
    u(end,j+1) = u(end,j) + 2*r*( ...
        u(end-1,j) - u(end,j) ...
        + h*(10 - 5*sin(pi*t(j)/T)));

end

% Value at x = 0.5, t = 0.1
ix = round(0.5/h) + 1;
jt = round(0.1/k) + 1;

fprintf('u(0.5,0.1) = %.6f\n',u(ix,jt));
```

```matlab

% Contour
figure();
contourf(x,t,u',25,'LineColor','none');
colormap(turbo);
colorbar;
xlabel('Position x');
ylabel('Time t');
title('FTCS - Neumann Boundary Conditions');
axis tight;
```

## Άσκηση 9.

Στα εσωτερικά σημεία εφαρμόζουμε τη μέθοδο FTCS:


$$u_i^{j+1} =u_i^j +r(u_{i-1}^j -2u_i^j +u_{i+1}^j )$$

Εφαρμόζουμε Forward Difference στην οριακή συνθήκη:


$$\begin{array}{l} \frac{u_2^{j+1} -u_1^{j+1} }{h}=4+6{\left(\frac{t_{j+1} }{T}\right)}^2 ewline \end{array}$$

Πολλαπλασιάζουμε με h:


$$\begin{array}{l} u_2^{j+1} -u_1^{j+1} =h\left\lbrack 4+6{\left(\frac{t_{j+1} }{T}\right)}^2 \right\rbrack ewline \end{array}$$

Λύνοντας ως προς την τιμή στο αριστερό άκρο, προκύπτει:


$$\begin{array}{l} u_1^{j+1} =u_2^{j+1} -h\left\lbrack 4+6{\left(\frac{t_{j+1} }{T}\right)}^2 \right\rbrack ewline \end{array}$$

Εφαρμόζουμε Backward Difference στην οριακή συνθήκη:


$$\begin{array}{l} \frac{u_{N_x }^{j+1} -u_{N_x -1}^{j+1} }{h}=-3e^{-t_{j+1} /T} ewline \end{array}$$

Πολλαπλασιάζουμε με h:


$$\begin{array}{l} u_{N_x }^{j+1} -u_{N_x -1}^{j+1} =-3he^{-t_{j+1} /T} ewline \end{array}$$

Λύνοντας ως προς την τιμή στο δεξί άκρο, προκύπτει:


$$\begin{array}{l} u_{N_x }^{j+1} =u_{N_x -1}^{j+1} -3he^{-t_{j+1} /T} ewline \end{array}$$

```matlab
clc; clear; close all;

L = 1;
T = 0.2;
h = 0.05;
k = 0.001;

x = (0:h:L)';
t = 0:k:T;

Nx = numel(x);
Nt = numel(t);

r = k/h^2;

assert(r <= 0.5, 'FTCS is unstable');

u = zeros(Nx,Nt);

% Initial condition
u(:,1) = 20 + 4*x - 3.5*x.^2 + 2*sin(pi*x).^2;

% FTCS method
i = 2:Nx-1;

for j = 1:Nt-1

    % Interior points
    u(i,j+1) = u(i,j) + r*( ...
        u(i-1,j) - 2*u(i,j) + u(i+1,j));

    % Forward Difference - Left boundary
    u(1,j+1) = u(2,j+1) ...
        - h*(4 + 6*(t(j+1)/T)^2);

    % Backward Difference - Right boundary
    u(end,j+1) = u(end-1,j+1) ...
        - 3*h*exp(-t(j+1)/T);

end

% Value at x = 0.5, t = 0.1
ix = round(0.5/h) + 1;
jt = round(0.1/k) + 1;

fprintf('u(0.5,0.1) = %.6f\n',u(ix,jt));
```

```matlab

% Contour
figure();
contourf(x,t,u',25,'LineColor','none');
colormap(turbo);
colorbar;
xlabel('Position x');
ylabel('Time t');
title('FTCS - Neumann Boundary Conditions');
axis tight;
```
