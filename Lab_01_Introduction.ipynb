### Conditionals (δομές επιλογής)

    if condition1:
        option1
    elif condition2:
        option2
    elif condition3:
        option3
    else:
        option4


Η Python εκτελεί το μπλοκ της πρώτης συνθήκης που αξιολογείται ως `True`. 

Αν καμία συνθήκη δεν ισχύει, εκτελείται το μπλοκ του `else`.


```python
air_temperature = 36
```

```python
if air_temperature < 10:
    print("It is cold!")
elif air_temperature > 30:
    print("It is hot!")
else:
    print("Normal conditions")
```

Η εσοχή (indent) (τέσσερα κενά) είναι **απαραίτητη** και ορίζει τα όρια του μπλοκ του conditional.

```python
a = 5.1
b = 10
c = 0

if (a > b) and (a > c):  # διπλή συνθήκη
    print("a is the largest")
else:  # το elif μπορεί να παραληφθεί
    print("a is not the largest")
```

- Slicing: επιλογή τμήματος της λίστας

        list[start:stop:step]

> Το stop δεν περιλαμβάνεται (exclusive).

> Το προεπιλεγμένο βήμα (step) είναι 1 και μπορεί να παραληφθεί.

```python
c[:3]  # αν παραληφθεί το start, ξεκινάμε από τη θέση 0
```

```python
c[1:]  # αν παραληφθεί το end, φτάνουμε ως το τέλος της λίστας
```

#### In-place methods

Αυτού του είδους οι μέθοδοι τροποποιούν απευθείας το αντικείμενο στη μνήμη του υπολογιστή.

Σε αντίθεση με τις κοινές μεθόδους, στις in-place δεν δημιουργούμε νέα μεταβλητή.

```python
numbers = [3, 2]
numbers.append(100)  # είναι λάθος να γράψουμε numbers = numbers.append(100)
numbers
```

#### Άσκηση: 

Έχουμε τις λίστες `[3, 2]` και `["Athens", "Berlin"]`

Στόχος είναι να ενώσετε την πρώτη λίστα με τη δεύτερη.

Χρησιμοποιήστε αρχικά τη μέθοδο `append` και έπειτα τη μέθοδο `extend`.

Προσοχή: και οι δύο παραπάνω μέθοδοι λειτουργούν in-place.

Συγκρίνετε τα δύο αποτελέσματα και παρατηρήστε τη διαφορά.

### Ορισμός μιας νέας συνάρτησης

```python
# εδώ ορίζουμε τη συνάρτηση

def add_two_numbers(a, b):
    c = a + b
    return c
```

Τα `a` και `b` είναι οι παράμετροι (parameters) της συνάρτησης.

```python
# εδώ χρησιμοποιούμε τη συνάρτηση

x = add_two_numbers(a=3, b=4)  # τα 3, 4 είναι τα arguments της συνάρτησης 
x
```

```python
y = add_two_numbers(10, 5)  # μπορούμε να παραλείψουμε τα ονόματα των παραμέτρων a, b
y
```
