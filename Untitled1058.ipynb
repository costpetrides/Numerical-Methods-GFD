```python
import numpy as np
import matplotlib.pyplot as plt

# Grid
x = np.linspace(0, 1, 6)

# Example function
u = x**2

# Grid spacing
h = x[1] - x[0]


# ==========================================
# 1. Vectorized slicing
# ==========================================

du_slice = (u[1:] - u[:-1]) / h

x_slice = x[:-1]


# ==========================================
# 2. Explicit indexing
# ==========================================

# Forward difference uses:
# u_i and u_{i+1}
#
# Therefore:
# L = 0
# R = 1

i = np.arange(0, len(u)-1)

du_index = (u[i+1] - u[i]) / h

x_index = x[i]


# ==========================================
# Results
# ==========================================

print("Using slicing:")
print(du_slice)

print("\nUsing explicit indexing:")
print(du_index)


# ==========================================
# Plot
# ==========================================

plt.plot(x_slice, du_slice, "o-", label="Slicing")
plt.plot(x_index, du_index, "x--", label="Explicit indexing")

plt.xlabel("x")
plt.ylabel("du/dx")
plt.legend()
plt.show()
```
