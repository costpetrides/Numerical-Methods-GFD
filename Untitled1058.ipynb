```python

import numpy as np
import matplotlib.pyplot as plt

# Grid
x = np.linspace(0, 1, 6)

# Example function
u = x**3

# Grid spacing
h = x[1] - x[0]

# Forward Difference
i = np.arange(0, len(u)-1)
x_index = x[i]
du_index = (u[i+1] - u[i]) / h


# Results
print('Forward Difference')
for x, du in zip(x_index, du_index): print(f"({x:.2f}, {du:.2f})")

# Plot
plt.plot(x_index, du_index, "o-", label="Forward Difference")

plt.xlabel("x")
plt.ylabel("du/dx")
plt.legend()
plt.grid(True)
plt.show()

print(type(u))
```
