```python
import numpy as np
import matplotlib.pyplot as plt

# Grid
x = np.arange(0, 6.2, 0.3 )
h = x[1] - x[0]
u = x**3
du=3*x**2

# Forward Difference
i_f = np.arange(0, len(u)-1)
x_f = x[i_f]
du_f = (u[i_f+1] - u[i_f]) / h

# Backward Difference
i_b = np.arange(1, len(u))
x_b = x[i_b]
du_b = (u[i_b] - u[i_b-1]) / h

# Central Difference
i_c = np.arange(1, len(u)-1)
x_c = x[i_c]
du_c = (u[i_c+1] - u[i_c-1]) / (2*h)


# Plot
plt.plot(x_f, du_f, "o-", label="Forward Difference")
plt.plot(x_b, du_b, "s-", label="Backward Difference")
plt.plot(x_c, du_c, "^-", label="Central Difference")
plt.plot(x, du, "--", label='Analitical' )
plt.xlabel("x")
plt.ylabel("du/dx")
plt.legend()
plt.grid(True)
plt.show()





# Results
print("Forward Difference")
for x_val, df in zip(x_f, du_f):
    print(f"({x_val:.2f}, {df:.2f})")

print("\nBackward Difference")
for x_val, db in zip(x_b, du_b):
    print(f"({x_val:.2f}, {db:.2f})")

print("\nCentral Difference")
for x_val, dc in zip(x_c, du_c):
    print(f"({x_val:.2f}, {dc:.2f})")

print(h)
```
