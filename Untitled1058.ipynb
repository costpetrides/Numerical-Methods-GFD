```python
import numpy as np
import matplotlib.pyplot as plt

# Grid
h = 0.5
x = np.arange(0, 10 + h, h)
u = x**3
du=3*x**2

# Forward Difference
i_f = np.arange(0, len(u)-1)
x_f = x[i_f]
du_f = (u[i_f+1] - u[i_f]) / h

# Backward Difference
i_b = np.arange(1, len(u))
x_b = x[i_b]
du_b = (u[i_b] - u[i_b-1]) / h

# Central Difference
i_c = np.arange(1, len(u)-1)
x_c = x[i_c]
du_c = (u[i_c+1] - u[i_c-1]) / (2*h)


# Plot
plt.plot(x_f, du_f, "-", label="Forward Difference")
plt.plot(x_b, du_b, "-", label="Backward Difference")
plt.plot(x_c, du_c, "-", label="Central Difference")
plt.plot(x, du, "--", label='Analitical' )
plt.xlabel("x")
plt.ylabel("du/dx")
plt.legend()
plt.grid(True)
plt.show()
```
