Η βιβλιοθήκη NumPy: 
- δίνει τη δυνατότητα για τη δημιουργία πινάκων (arrays)

- προσφέρει μεγάλη συλλογή μαθηματικών συναρτήσεων

Η NumPy δεν περιλαμβάνεται στις βιβλιοθήκες που φορτώνονται αυτόματα, οπότε πρέπει να την εισάγουμε.

Ο παρακάτω τρόπος είναι ο καθιερωμένος για εισαγωγή της, χρησιμοποιώντας τη συντομογραφία `np`.

```python
# εισαγωγή της βιβλιοθήκης NumPy

import numpy as np
```

Θέλουμε να προσθέσουμε δύο λίστες μεταξύ τους, στοιχείο με στοιχείο

Ας δοκιμάσουμε αρχικά χωρίς να χρησιμοποιήσουμε τη βιβλιοθήκη NumPy:

```python
list_a = [3, 2, 1]
list_b = [2, 5, 6]
```

```python
# λάθος τρόπος Νο1

list_a + list_b  # το + απλά ενώνει δύο λίστες
```

```python
# "λάθος" τρόπος Νο2

list_c = [None, None, None]

for i in range(0, len(list_a), 1):
    list_c[i] = list_a[i] + list_b[i]
list_c
```

Χρησιμοποιώντας τη βιβλιοθήκη NumPy:

```python
a = np.array(list_a)  # μέσω της συνάρτησης array φτιάχουμε ένα ndarray object
a
```

```python
type(a)
```

```python
b = np.array(list_b)
b
```

```python
c = a + b  # η πράξη γίνεται ανά στοιχείο (element-wise)
c
```

```python
d = a * b  # η πράξη γίνεται ανά στοιχείο
d
```

### Numpy arange

Θυμηθείτε από το πρώτο μάθημα τη συνάρτηση `range`:

    range(start, stop, step)

Η βιβλιοθήκη NumPy παρέχει τη δική της εκδοχή:

    np.arange(start, stop, step)

> Stop is exclusive (δηλαδή το στοιχείο σε αυτή τη θέση δεν συμπεριλαμβάνεται)

> Προσοχή: το `np.arange` γράφεται με ένα μόνο "r" σε αντίθεση με τη λέξη *arrange*

```python
np.arange(3, 12, 1)
```

Για να συμπεριλάβουμε και το 12, μπορούμε να γράψουμε:

```python
np.arange(3, 12 + 1, 1)  # ή φυσικά np.arange(3, 13, 1)
```

### Συναρτήσεις

- Έλεγχος ισότητας

```python
a = np.array([5, 3, 2])
b = np.array([5, 3, 2])
```

```python
np.equal(a, b)  # σύγκριση στοιχείο με στοιχείο 
```

```python
np.array_equal(a, b)  # συνολική σύγκριση
```

```python
np.allclose(a, b)  # συνολική σύγκριση για αριθμούς με πολλά δεκαδικά στοιχεία

# relative tolerance: rtol=1e-05
# absolute tolerance: atol=1e-08
```

### Δημιουργία διαγωνίου πίνακα

\begin{pmatrix}
10 & 0 & 0 \\[3pt]
0 & 10 & 0 \\[3pt]
0 & 0 & 10
\end{pmatrix}



```python
a = np.diag(10 * np.ones(3))
a
```

#### Αναλυτικά τα βήματα

Το `np.ones(3)` δημιουργεί έναν μονοδιάστατο πίνακα με τρία στοιχεία όλα ίσα με 1:


```python
np.ones(3)
```

Το πολλαπλασιάζουμε με 10 ώστε κάθε στοιχείο να γίνει ίσο με 10:


```python
10 * np.ones(3)
```

Η συνάρτηση `np.diag()` χρησιμοποιεί το `10 * np.ones(3)` ως παράμετρο και δημιουργεί έναν νέο πίνακα, τοποθετώντας αυτά τα στοιχεία στη διαγώνιο του.

```python
np.diag(10 * np.ones(3))
```

Αν θέλουμε η διαγώνιος να είναι μετατοπισμένη, τότε χρησιμοποιούμε τη δεύτερη παράμετρο του `np.diag()`, η οποία μπορεί να λάβει τιμές `1` ή `-1`.


```python
np.diag(10 * np.ones(2), 1)  # πάνω από την κύρια διαγώνιο
```

```python
np.diag(10 * np.ones(2), -1)  # κάτω από την κύρια διαγώνιο
```

> Προσοχή: για να κρατήσουμε τις διαστάσεις του πίνακα 3 × 3, το np.ones χρειάζεται 2 στοιχεία για τις μετατοπισμένες διαγώνιες.
