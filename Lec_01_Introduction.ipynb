Οι περισσότερες διεργασίες που μελετούν οι φυσικές επιστήμες περιγράφονται από διαφορικές εξισώσεις:

- Αριθμητικά μοντέλα πρόγνωσης καιρού

- Κλιματικές προσομοιώσεις

- Ωκεάνια κυκλοφορία

- Μεταφορά ατμοσφαιρικών ρύπων

- Ροή αέρα γύρω από κτίρια


Κοινό χαρακτηριστικό όλων των παραπάνω είναι ότι η εξέλιξη του φυσικού συστήματος περιγράφεται μέσω παραγώγων ως προς τον χώρο και τον χρόνο.

Σκοπός των αριθμητικών μεθόδων είναι να μετατρέψουν αυτές τις μαθηματικές εξισώσεις σε μορφή που μπορεί να επιλυθεί υπολογιστικά, με όσο το δυνατόν μεγαλύτερη ακρίβεια και μικρότερο υπολογιστικό κόστος.

### Γιατί χρειαζόμαστε αριθμητικές μεθόδους;

### 1. Οι περισσότερες διαφορικές εξισώσεις δεν επιλύονται αναλυτικά

Οι διαφορικές εξισώσεις αποτελούν το βασικό εργαλείο περιγραφής των φυσικών διεργασιών. 

Σε ορισμένες απλές περιπτώσεις είναι δυνατή η εύρεση μιας αναλυτικής λύσης, δηλαδή μιας ακριβούς μαθηματικής έκφρασης για τη λύση.

Στα περισσότερα όμως πραγματικά προβλήματα η αναλυτική λύση είτε δεν υπάρχει είτε είναι πρακτικά εξαιρετικά δύσκολο να βρεθεί.

Οι κυριότεροι λόγοι είναι:

- μη γραμμικότητα των εξισώσεων

- πολύπλοκες αρχικές και οριακές συνθήκες

- πολύπλοκη γεωμετρία του προβλήματος

- σύζευξη πολλών φυσικών διεργασιών

Για τον λόγο αυτό, η επίλυση των περισσότερων πραγματικών προβλημάτων βασίζεται σε αριθμητικές προσεγγίσεις και όχι σε αναλυτικές λύσεις.

### 2. Οι υπολογιστές λειτουργούν με διακριτά μαθηματικά

Στις διαφορικές εξισώσεις οι μεταβλητές θεωρούνται συνεχείς συναρτήσεις του χώρου και του χρόνου, ενώ επίσης οι παράγωγοι ορίζονται μέσω της έννοιας του ορίου.

Ένας υπολογιστής αντίθετα λειτουργεί με διακριτές και πεπερασμένες αναπαραστάσεις. Μπορεί να αποθηκεύσει πεπερασμένο πλήθος αριθμών και να αναπαραστήσει τους πραγματικούς αριθμούς με πεπερασμένη ακρίβεια.

Κατά συνέπεια, οι συνεχείς μαθηματικές έννοιες, όπως τα όρια και οι παράγωγοι, δεν μπορούν να αναπαρασταθούν επακριβώς από έναν υπολογιστή.

Για να επιλύσουμε λοιπόν μια διαφορική εξίσωση πρέπει να μετατρέψουμε το συνεχές μαθηματικό πρόβλημα σε ένα διακριτό αριθμητικό πρόβλημα που μπορεί να επιλυθεί υπολογιστικά.

Η διαδικασία αυτή ονομάζεται διακριτοποίηση (discretization).

### Γιατί ο υπολογιστής χρησιμοποιεί το δυαδικό σύστημα;

Οι υπολογιστές αποτελούνται από ψηφιακά ηλεκτρονικά κυκλώματα, τα οποία περιλαμβάνουν δισεκατομμύρια μικροσκοπικά τρανζίστορ που λειτουργούν ως ηλεκτρονικοί διακόπτες.

Η λειτουργία των ψηφιακών κυκλωμάτων βασίζεται στη διάκριση μεταξύ δύο λογικών καταστάσεων, οι οποίες συμβολίζονται ως:

$$
0 \quad \text{και} \quad 1
$$

Κάθε μία από αυτές τις τιμές αποτελεί ένα δυαδικό ψηφίο (binary digit ή bit).

Κατά συνέπεια, κάθε πληροφορία που αποθηκεύεται ή επεξεργάζεται από έναν υπολογιστή, όπως για παράδειγμα ένας αριθμός, μια εικόνα ή μια εντολή προγράμματος, αναπαρίσταται τελικά ως μια ακολουθία δυαδικών ψηφίων.

Η συνάρτηση `bin()` της Python μας δίνει τη δυαδική αναπαράσταση ενός ακέραιου αριθμού:

```python
bin(2)
```

Το πρόθεμα **`0b`** δηλώνει ότι ο αριθμός που ακολουθεί είναι γραμμένος σε δυαδική μορφή (binary).

Φυσικά, μπορούμε να πραγματοποιούμε αριθμητικές πράξεις χρησιμοποιώντας το γνωστό δεκαδικό σύστημα, χωρίς να χρειάζεται να γνωρίζουμε τη δυαδική αναπαράσταση.

```python
1 + 2
```

```python
0.1 + 0.2
```

### Αναπαράσταση ακέραιων και πραγματικών αριθμών

Οι ακέραιοι αριθμοί μπορούν να αναπαρασταθούν ακριβώς στο δυαδικό σύστημα.

Για παράδειγμα:

$$
5_{10}=101_2
$$

και επομένως ο αριθμός 5 μπορεί να αποθηκευτεί χωρίς σφάλμα.

Στους πραγματικούς αριθμούς, όμως, η κατάσταση είναι διαφορετική. Ορισμένοι αριθμοί μπορούν να αναπαρασταθούν ακριβώς στο δυαδικό σύστημα, ενώ άλλοι όχι.

Για παράδειγμα:

$$
0.5_{10}=0.1_2
$$

$$
0.25_{10}=0.01_2
$$

$$
0.75_{10}=0.11_2
$$

Οι αριθμοί αυτοί αναπαρίστανται ακριβώς, επειδή μπορούν να εκφραστούν ως πεπερασμένα αθροίσματα αρνητικών δυνάμεων του 2:

$$
0.75 = \frac{1}{2} + \frac{1}{4} = 2^{-1} + 2^{-2}
$$

Αντίθετα, πολλοί δεκαδικοί αριθμοί δεν έχουν πεπερασμένη δυαδική αναπαράσταση. Για παράδειγμα:

$$
0.1_{10}=0.0001100110011\ldots
$$

Η δυαδική αναπαράσταση του 0.1 συνεχίζεται επ' άπειρον. Επειδή ο υπολογιστής διαθέτει πεπερασμένο αριθμό bits, αποθηκεύει την πλησιέστερη δυνατή δυαδική προσέγγιση.

Το ίδιο συμβαίνει και με πολλούς άλλους αριθμούς, όπως:

$$
0.2,\quad 0.3,\quad 0.6,\quad 0.7
$$

Η μικρή διαφορά μεταξύ της πραγματικής τιμής ενός αριθμού και της τιμής που μπορεί να αναπαρασταθεί στον υπολογιστή αποτελεί σφάλμα στρογγυλοποίησης (round-off error).

Το σφάλμα είναι συνήθως εξαιρετικά μικρό, ωστόσο, μπορεί να συσσωρευτεί μετά από μεγάλο αριθμό αριθμητικών πράξεων και να γίνει σημαντικό σε υπολογισμούς που είναι ιδιαίτερα ευαίσθητοι σε μικρές αριθμητικές αποκλίσεις.

Επιπλέον, οι μικρές αυτές αποκλίσεις μπορούν να προκαλέσουν μη αναμενόμενα αποτελέσματα σε ελέγχους ακριβούς ισότητας. 

Για παράδειγμα, μια τιμή που θεωρητικά θα έπρεπε να είναι ακριβώς ίση με 3 μπορεί στον υπολογιστή να είναι ελάχιστα μεγαλύτερη ή μικρότερη από 3.

Καταλήγοντας, οι πραγματικοί αριθμοί αποθηκεύονται στον υπολογιστή χρησιμοποιώντας floating-point representation (αναπαράσταση κινητής υποδιαστολής), δηλαδή μια πεπερασμένη δυαδική αναπαράσταση της αριθμητικής τους τιμής. 

Τα σφάλματα στρογγυλοποίησης αποτελούν συνεπώς εγγενές χαρακτηριστικό των αριθμητικών υπολογισμών με πεπερασμένη ακρίβεια.


```python
print(format(0.5, ".20f"))

print(format(0.1, ".20f"))

print(0.1 + 0.2)
```

### Πεπερασμένες διαφορές

### Τι εκφράζει η πρώτη παράγωγος;

Η πρώτη παράγωγος εκφράζει τον τοπικό ρυθμό μεταβολής μιας συνάρτησης ως προς μια ανεξάρτητη μεταβλητή.

Πιο απλά, περιγράφει πόσο μεταβάλλεται μια φυσική ποσότητα ανά μονάδα μεταβολής μιας άλλης, σε ένα συγκεκριμένο σημείο.

Η έννοια αυτή αποτελεί θεμελιώδες εργαλείο στη μαθηματική περιγραφή φυσικών διεργασιών, καθώς οι περισσότεροι φυσικοί νόμοι δεν περιγράφουν τις ίδιες τις ποσότητες, αλλά τον τρόπο με τον οποίο αυτές μεταβάλλονται στον χώρο και στον χρόνο.

### Πώς υπολογίζει ένας υπολογιστής μια παράγωγο;

Ο μαθηματικός ορισμός της παραγώγου βασίζεται στην έννοια του ορίου:

$$
\frac{du}{dx}
=
\lim_{h\rightarrow0}
\frac{u(x+h)-u(x)}{h}
$$

Ωστόσο, ένας υπολογιστής δεν μπορεί να υπολογίσει ένα όριο, καθώς δεν μπορεί να χρησιμοποιήσει μία απειροστά μικρή μεταβολή ($h \rightarrow 0$).

Αντίθετα, μπορεί να χρησιμοποιήσει μόνο μία πεπερασμένη μεταβολή της ανεξάρτητης μεταβλητής.

Η βασική ιδέα των αριθμητικών μεθόδων είναι επομένως η αντικατάσταση της παραγώγου από μία αριθμητική προσέγγιση, χρησιμοποιώντας ένα μικρό αλλά πεπερασμένο βήμα.

Η διαδικασία αυτή είναι γνωστή ως **πεπερασμένες διαφορές** (finite differences).

### Forward finite difference

Αφού ο υπολογιστής δεν μπορεί να χρησιμοποιήσει το όριο $h\rightarrow0$ για τον υπολογισμό της παραγώγου στο σημείο $x$, τότε η φυσική επιλογή είναι να χρησιμοποιήσει ένα μικρό αλλά πεπερασμένο βήμα.

Η απλούστερη προσέγγιση είναι η χρήση του επόμενου διαθέσιμου σημείου.

Η προσέγγιση αυτή ονομάζεται forward finite difference και δίνεται από:

$$
\frac{du}{dx}
\approx
\frac{u(x+\Delta x)-u(x)}{\Delta x}
$$

### Παράδειγμα υπολογισμού

Έστω ότι μία φυσική ποσότητα έχει μετρηθεί στα ακόλουθα χωρικά σημεία:

| | $x_0$ | $x_1$ | $x_2$ | $x_3$ |
|----:|--:|--:|--:|--:|
| $x$ | 0 | 1 | 2 | 3 |
| $u(x)$ | 10 | 12 | 15 | 19 |

Η απόσταση μεταξύ διαδοχικών σημείων είναι:

$$
\Delta x = 1
$$

Να υπολογιστεί η πρώτη παράγωγος στο σημείο $x=1$ χρησιμοποιώντας τη μέθοδο forward finite difference.

```python
u = np.array([10.0, 12.0, 15.0, 19.0])
dx = 1.0

du_dx_forward = (u[2] - u[1]) / dx

print(du_dx_forward)
```

### Backward finite difference

Αντί να χρησιμοποιήσουμε το επόμενο σημείο, μπορούμε να χρησιμοποιήσουμε το προηγούμενο.

Η προσέγγιση αυτή ονομάζεται backward finite difference και δίνεται από:

$$
\frac{du}{dx}
\approx
\frac{u(x)-u(x-\Delta x)}{\Delta x}
$$

```python
du_dx_backward = (u[1] - u[0]) / dx

print(du_dx_backward)
```

### Centered finite difference

Μια καλύτερη ιδέα είναι να χρησιμοποιήσουμε πληροφορία και από τις δύο πλευρές του σημείου ενδιαφέροντος.

Η προσέγγιση αυτή ονομάζεται centered finite difference και δίνεται από:

$$
\frac{du}{dx}
\approx
\frac{u(x+\Delta x)-u(x-\Delta x)}
{2\Delta x}
$$

```python
du_dx_central = (u[2] - u[0]) / (2 * dx)

print(du_dx_central)
```

Η centered finite difference μπορεί να θεωρηθεί ως ο μέσος όρος των forward και backward finite differences:

$$
\frac{1}{2}
\left(
\frac{u(x+\Delta x)-u(x)}{\Delta x}
+
\frac{u(x)-u(x-\Delta x)}{\Delta x}
\right)
=
\frac{u(x+\Delta x)-u(x-\Delta x)}
{2\Delta x}
$$

Ο συμμετρικός αυτός υπολογισμός οδηγεί, όπως θα αποδειχθεί μέσω της ανάπτυξης Taylor στις επόμενες διαλέξεις, σε μεγαλύτερη ακρίβεια από τις forward και backward προσεγγίσεις.

```python
(du_dx_forward + du_dx_backward) / 2
```

### Τι εκφράζει η δεύτερη παράγωγος;

Η δεύτερη παράγωγος εκφράζει τον ρυθμό μεταβολής της πρώτης παραγώγου.

Με άλλα λόγια, περιγράφει πώς μεταβάλλεται ο ίδιος ο ρυθμός μεταβολής μιας φυσικής ποσότητας στον χώρο ή στον χρόνο.

Η δεύτερη παράγωγος εμφανίζεται σε πολλές φυσικές διεργασίες, όπως η διάχυση θερμότητας, η διάχυση μάζας και η διάδοση ορμής.

### Φυσική ερμηνεία της δεύτερης παραγώγου

Στις διεργασίες διάχυσης, η πρώτη χωρική παράγωγος συνδέεται με την πυκνότητα ροής (flux) μιας φυσικής ποσότητας.

$$
q = -k \frac{\partial T}{\partial x}
$$

Στην περίπτωση της θερμότητας, μια χωρική μεταβολή της θερμοκρασίας δημιουργεί ροή θερμότητας από τις θερμότερες προς τις ψυχρότερες περιοχές.

Η δεύτερη χωρική παράγωγος περιγράφει πώς μεταβάλλεται αυτή η ροή από σημείο σε σημείο.

Για ένα συγκεκριμένο σημείο:

- αν η εισερχόμενη ροή θερμότητας είναι μεγαλύτερη από την εξερχόμενη, συσσωρεύεται θερμότητα και η θερμοκρασία αυξάνεται

- αν η εξερχόμενη ροή θερμότητας είναι μεγαλύτερη από την εισερχόμενη, απομακρύνεται θερμότητα και η θερμοκρασία μειώνεται

Η φυσική αυτή διαδικασία περιγράφεται από την εξίσωση διάδοσης της θερμότητας:

$$
\frac{\partial T}{\partial t}
=
\alpha
\frac{\partial^2 T}{\partial x^2}
$$

όπου $\alpha$ είναι ο συντελεστής θερμικής διάχυσης.

Η εξίσωση δηλώνει ότι ο ρυθμός μεταβολής της θερμοκρασίας στον χρόνο εξαρτάται από τη δεύτερη χωρική παράγωγο της θερμοκρασίας.

Επομένως:

- Διαφορές στη θερμοκρασία → πυκνότητα ροής θερμότητας (πρώτη χωρική παράγωγος)

- Διαφορές στη ροή θερμότητας → μεταβολή της θερμοκρασίας (δεύτερη χωρική παράγωγος)

Η αριθμητική επίλυση της εξίσωσης διάδοσης της θερμότητας θα αποτελέσει βασικό αντικείμενο των επόμενων διαλέξεων.

### Προσέγγιση της δεύτερης παραγώγου

Με τον ίδιο τρόπο που προσεγγίσαμε την πρώτη παράγωγο, μπορούμε να προσεγγίσουμε και τη δεύτερη παράγωγο χρησιμοποιώντας τις τιμές της συνάρτησης σε γειτονικά σημεία.

Χωρίς προς το παρόν να την αποδείξουμε, η συνηθέστερη αριθμητική προσέγγιση είναι:

$$
\frac{d^2u}{dx^2}
\approx
\frac{u(x-\Delta x)-2u(x)+u(x+\Delta x)}
{\Delta x^2}
$$

Η μαθηματική απόδειξη της παραπάνω σχέσης θα παρουσιαστεί στην επόμενη διάλεξη μέσω της ανάπτυξης Taylor.

Για μια πιο διαισθητική ερμηνεία μπορούμε να αναφέρουμε τα εξής:

Η παράσταση

$$
\frac{u(x-\Delta x)+u(x+\Delta x)}{2}
$$

εκφράζει τον μέσο όρο της συνάρτησης στα δύο γειτονικά σημεία του $x$.

Η δεύτερη παράγωγος συγκρίνει την τιμή της συνάρτησης στο σημείο $x$ με αυτόν τον τοπικό μέσο όρο.

Συγκεκριμένα η ποσότητα:

$$
\frac{u(x-\Delta x)+u(x+\Delta x)}{2}-u(x)
$$

εκφράζει πόσο διαφέρει η τιμή της συνάρτησης από τον μέσο όρο των γειτονικών σημείων.

Παρατηρούμε ότι:

$$
\frac{u(x-\Delta x)-2u(x)+u(x+\Delta x)}
{\Delta x^2}
=
\frac{2}{\Delta x^2}
\left(
\frac{u(x-\Delta x)+u(x+\Delta x)}{2}
-
u(x)
\right)
$$

Αν η τιμή ενός σημείου είναι μικρότερη από τον μέσο όρο των γειτονικών του σημείων, τότε η δεύτερη παράγωγος είναι θετική και αναμένουμε η τιμή του σημείου να αυξηθεί με τον χρόνο στα φαινόμενα διάχυσης. 

Αντίθετα, όταν είναι μεγαλύτερη από τον τοπικό μέσο όρο, αναμένουμε να μειωθεί.

```python
u = np.array([10., 12., 15., 19., 24.])
dx = 1.0

average = (u[1] + u[3]) / 2
difference = average - u[2]

print(average)
print(difference)
```

### Κατηγορίες Μερικών Διαφορικών Εξισώσεων

Οι γραμμικές Μερικές Διαφορικές Εξισώσεις (ΜΔΕ) (Partial Differential Equations, PDEs) δεύτερης τάξης μπορούν να ταξινομηθούν σε τρεις βασικές κατηγορίες:

- παραβολικές (parabolic)

- ελλειπτικές (elliptic)

- υπερβολικές (hyperbolic)

### Μαθηματική ταξινόμηση

Θεωρούμε μία ΜΔΕ δεύτερης τάξης δύο ανεξάρτητων μεταβλητών της μορφής:

$$
\alpha\frac{\partial^2u}{\partial x^2}
+
\beta\frac{\partial^2u}{\partial x\partial y}
+
\gamma\frac{\partial^2u}{\partial y^2}
=
\Psi
$$

Η ταξινόμηση της εξίσωσης καθορίζεται από την ποσότητα:

$$
\beta^2-4\alpha\gamma
$$

Ανάλογα με το πρόσημό της:

$$
\begin{aligned}
\beta^2-4\alpha\gamma &>0
&&\Rightarrow \text{υπερβολική}\\[5pt]
\beta^2-4\alpha\gamma &=0
&&\Rightarrow \text{παραβολική}\\[5pt]
\beta^2-4\alpha\gamma &<0
&&\Rightarrow \text{ελλειπτική}
\end{aligned}
$$

Οι τρεις κατηγορίες έχουν διαφορετικές μαθηματικές ιδιότητες και αντιστοιχούν σε διαφορετικού τύπου φυσικά προβλήματα.

### Παραβολικές εξισώσεις

Χαρακτηριστικό παράδειγμα παραβολικής ΜΔΕ είναι η εξίσωση της θερμότητας:

$$
\frac{\partial u}{\partial t}
=
α\frac{\partial^2u}{\partial x^2}
$$

όπου $α$ ο συντελεστής θερμικής διάχυσης.

Οι παραβολικές εξισώσεις περιγράφουν συνήθως διαδικασίες διάχυσης και εξομάλυνσης ενός πεδίου με την πάροδο του χρόνου.

Στην εξίσωση της θερμότητας, μία αρχική κατανομή θερμοκρασίας εξελίσσεται χρονικά καθώς η θερμότητα διαδίδεται από περιοχές υψηλότερης προς περιοχές χαμηλότερης θερμοκρασίας.

Για την επίλυση ενός τέτοιου προβλήματος απαιτείται συνήθως μία αρχική συνθήκη και κατάλληλες οριακές συνθήκες.

### Υπερβολικές εξισώσεις

Χαρακτηριστικό παράδειγμα υπερβολικής ΜΔΕ είναι η εξίσωση κύματος:

$$
\frac{\partial^2u}{\partial t^2}
=
c^2\frac{\partial^2u}{\partial x^2}
$$

Οι υπερβολικές εξισώσεις περιγράφουν συνήθως φαινόμενα στα οποία η πληροφορία διαδίδεται μέσα στον χώρο.

Στην εξίσωση κύματος, η διάδοση πραγματοποιείται κατά μήκος χαρακτηριστικών καμπυλών:

$$
x-ct=\text{σταθερά}
$$

$$
x+ct=\text{σταθερά}
$$

Επομένως, μία διαταραχή μεταδίδεται με πεπερασμένη ταχύτητα $c$.

### Ελλειπτικές εξισώσεις

Χαρακτηριστικό παράδειγμα ελλειπτικής ΜΔΕ είναι η εξίσωση Laplace:

$$
\frac{\partial^2u}{\partial x^2}
+
\frac{\partial^2u}{\partial y^2}
=
0
$$

Οι ελλειπτικές εξισώσεις περιγράφουν συνήθως καταστάσεις ισορροπίας, χωρίς χρονική εξέλιξη.

Η λύση σε ένα σημείο εξαρτάται από τις συνθήκες στο συνολικό όριο του πεδίου. Για τον λόγο αυτό οι ελλειπτικές εξισώσεις αντιμετωπίζονται συνήθως ως προβλήματα οριακών τιμών.

Ένα χαρακτηριστικό φυσικό παράδειγμα είναι η κατανομή θερμοκρασίας σε μία περιοχή όταν το σύστημα έχει φτάσει σε θερμική ισορροπία.
