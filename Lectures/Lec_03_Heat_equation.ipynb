## Εξίσωση διάδοσης θερμότητας
________


Η μονοδιάστατη εξίσωση διάδοσης της θερμότητας γράφεται

$$
\frac{\partial u}{\partial t}
=
α\frac{\partial^2u}{\partial x^2}
$$

όπου:

- $u(x,t)$ η θερμοκρασία

- $α$ ο συντελεστής θερμικής διάχυσης

Για λόγους απλότητας, θεωρούμε αρχικά

$$
α=1
$$

οπότε η εξίσωση γράφεται ως:

$$
\frac{\partial u}{\partial t}
=
\frac{\partial^2u}{\partial x^2}
$$

Ο πρώτος στόχος μας είναι να αντικαταστήσουμε τις παραγώγους με πεπερασμένες διαφορές, ώστε να μπορεί η εξίσωση να λυθεί αριθμητικά.

### Αρχικές και οριακές συνθήκες

Για την αριθμητική επίλυση της εξίσωσης απαιτείται να γνωρίζουμε την αρχική κατανομή της θερμοκρασίας, καθώς και τις οριακές συνθήκες στα άκρα του χωρικού διαστήματος.

Στη συνέχεια επικεντρωνόμαστε στη διακριτοποίηση της εξίσωσης για τα εσωτερικά σημεία του χωρικού πλέγματος.

### Διακριτοποίηση του χώρου και του χρόνου

Χωρίζουμε τον χώρο σε σημεία με σταθερό βήμα

$$
\Delta x=h
$$

και τον χρόνο σε χρονικά βήματα

$$
\Delta t=k
$$

Οι συντεταγμένες κάθε σημείου γράφονται

$$
x_i=ih
$$

$$
t_j=jk
$$

Η αριθμητική λύση στο σημείο $(x_i,t_j)$ συμβολίζεται ως

$$
u_{i,j}
$$

Ο δείκτης:

- $i$ αντιστοιχεί στη θέση στον χώρο

- $j$ αντιστοιχεί στη χρονική στιγμή

Στη συνέχεια παρουσιάζονται τρία βασικά σχήματα πεπερασμένων διαφορών για την αριθμητική επίλυση της εξίσωσης θερμότητας.

## 1. Explicit

### Forward Time Centered Space (FTCS)

Η εξίσωση της θερμότητας μας δίνει τον ρυθμό μεταβολής της θερμοκρασίας ως προς τον χρόνο.

Για να υπολογίσουμε τη θερμοκρασία στην επόμενη χρονική στιγμή, αντικαθιστούμε:

- τη χρονική παράγωγο με forward finite difference

- τη δεύτερη χωρική παράγωγο με centered finite difference.

Η διακριτή μορφή της εξίσωσης θα μας επιτρέψει να υπολογίσουμε διαδοχικά τη θερμοκρασία σε κάθε νέο χρονικό βήμα.

### Διακριτοποίηση της χρονικής παραγώγου

Η χρονική παράγωγος προσεγγίζεται ως

$$
\frac{\partial u}{\partial t}
\approx
\frac{u_{i,j+1}-u_{i,j}}{k}
$$

Η παράγωγος υπολογίζεται στη χρονική στιγμή $t_j$, χρησιμοποιώντας forward finite difference.

### Διακριτοποίηση της χωρικής παραγώγου

Η δεύτερη χωρική παράγωγος προσεγγίζεται ως

$$
\frac{\partial^2u}{\partial x^2}
\approx
\frac{u_{i-1,j}-2u_{i,j}+u_{i+1,j}}{h^2}
$$

Η παράγωγος υπολογίζεται επίσης στη χρονική στιγμή $t_j$, χρησιμοποιώντας centered finite difference.

Αντικαθιστώντας τις δύο προσεγγίσεις στην εξίσωση

$$
\frac{\partial u}{\partial t}
=
\frac{\partial^2u}{\partial x^2}
$$

προκύπτει

$$
\frac{u_{i,j+1}-u_{i,j}}{k}
=
\frac{u_{i-1,j}-2u_{i,j}+u_{i+1,j}}{h^2}
$$

### Επίλυση ως προς την επόμενη χρονική στιγμή

Λύνοντας ως προς $u_{i,j+1}$

$$
u_{i,j+1}
=
u_{i,j}
+
\frac{k}{h^2}
\left(
u_{i-1,j}-2u_{i,j}+u_{i+1,j}
\right)
$$

Ορίζουμε τον συντελεστή

$$
r=\frac{k}{h^2}
$$

οπότε

$$
u_{i,j+1}
=
u_{i,j}
+
r
\left(
u_{i-1,j}-2u_{i,j}+u_{i+1,j}
\right)
$$

Αναπτύσσοντας τους όρους:

$$
u_{i,j+1}
=
ru_{i-1,j}
+
(1-2r)u_{i,j}
+
ru_{i+1,j}
$$

Η μέθοδος ονομάζεται ρητή (explicit), επειδή η θερμοκρασία στο επόμενο χρονικό βήμα υπολογίζεται αποκλειστικά από γνωστές θερμοκρασίες της τρέχουσας χρονικής στιγμής.


#### Ευστάθεια της μεθόδου FTCS

Η μέθοδος FTCS είναι **υπό συνθήκη ευσταθής** (conditionally stable), δηλαδή η ευστάθειά της εξαρτάται από τη σχέση μεταξύ του χρονικού βήματος $k$ και του χωρικού βήματος $h$.

Το κριτήριο ευστάθειας είναι το εξής:

$$
\boxed{r\leq\frac{1}{2}\quad\Rightarrow\quad k\leq\frac{h^2}{2}}
$$

Όταν η συνθήκη ικανοποιείται, τα αριθμητικά σφάλματα δεν ενισχύονται κατά την εξέλιξη της λύσης.

Αντίθετα, για $r>1/2$, ορισμένα σφάλματα μπορούν να ενισχύονται, προκαλώντας μη φυσικές ταλαντώσεις και αριθμητική αστάθεια.

Συνεπώς, αν μειώσουμε το χωρικό βήμα στο μισό, πρέπει να μειώσουμε το χρονικό βήμα τουλάχιστον στο ένα τέταρτο.

Για τη γενική εξίσωση θερμότητας με συντελεστή διάχυσης $\alpha$, το κριτήριο είναι:

$$
\boxed{\frac{\alpha k}{h^2}\leq\frac{1}{2}}
$$


## 2. Implicit

### Backward Time Centered Space (BTCS)

Στην explicit μέθοδο υπολογίσαμε τη νέα θερμοκρασία χρησιμοποιώντας τη χωρική παράγωγο στην τρέχουσα χρονική στιγμή.

Τώρα ακολουθούμε μία διαφορετική προσέγγιση, απαιτώντας η εξίσωση της θερμότητας να ικανοποιείται στην **επόμενη χρονική στιγμή**.

### Διακριτοποίηση της χρονικής παραγώγου

Η χρονική παράγωγος προσεγγίζεται με μία backward finite difference:

$$
\frac{\partial u}{\partial t}
\approx
\frac{u_{i,j+1}-u_{i,j}}{k}
$$

Η παράγωγος υπολογίζεται στη χρονική στιγμή $t_{j+1}$

### Διακριτοποίηση της χωρικής παραγώγου

Η δεύτερη χωρική παράγωγος υπολογίζεται επίσης στην επόμενη χρονική στιγμή $t_{j+1}$:

$$
\frac{\partial^2u}{\partial x^2}
\approx
\frac{u_{i-1,j+1}-2u_{i,j+1}+u_{i+1,j+1}}{h^2}
$$

Δηλαδή, χρησιμοποιούμε την άγνωστη θερμοκρασία της επόμενης χρονικής στιγμής για να υπολογίσουμε τη διάχυση της θερμότητας.

Αντικαθιστώντας τις δύο προσεγγίσεις,

$$
\frac{u_{i,j+1}-u_{i,j}}{k}
=
\frac{u_{i-1,j+1}-2u_{i,j+1}+u_{i+1,j+1}}{h^2}
$$

Θέτοντας πάλι

$$
r=\frac{k}{h^2}
$$

προκύπτει

$$
u_{i,j+1}-u_{i,j}
=
r\,u_{i-1,j+1}
-
2r\,u_{i,j+1}
+
r\,u_{i+1,j+1}
$$

Μεταφέροντας όλους τους άγνωστους όρους στο αριστερό μέλος:

$$
-r\,u_{i-1,j+1}
+
(1+2r)u_{i,j+1}
-
r\,u_{i+1,j+1}
=
u_{i,j}
$$

Η μέθοδος ονομάζεται *Implicit Backward Time Centered Space (BTCS)* επειδή:

- η χρονική παράγωγος προσεγγίζεται με backward finite difference

- η δεύτερη χωρική παράγωγος προσεγγίζεται με centered finite difference

- η εξίσωση της θερμότητας αντιστοιχεί στην επόμενη χρονική στιγμή.

Η εύρεση των θερμοκρασιών απαιτεί την επίλυση ενός συστήματος γραμμικών εξισώσεων.

## 3. Crank – Nicolson

Στην explicit μέθοδο η εξίσωση της θερμότητας αναφέρεται στην τρέχουσα χρονική στιγμή $t_j$.

Στην implicit μέθοδο αναφέρεται στην επόμενη χρονική στιγμή $t_{j+1}$.

Η μέθοδος Crank – Nicolson θεωρεί την εξίσωση στο μέσο του χρονικού διαστήματος

$$
t_{j+\frac{1}{2}}
=
t_j+\frac{\Delta t}{2}
$$

Η βασική ιδέα είναι να χρησιμοποιηθεί ο μέσος όρος της χωρικής παραγώγου στις χρονικές στιγμές $t_j$ και $t_{j+1}$.

### Διακριτοποίηση της χρονικής παραγώγου

Για την εξίσωση της θερμότητας

$$
\frac{\partial u}{\partial t}
=
\frac{\partial^2u}{\partial x^2}
$$

η μέθοδος Crank – Nicolson θεωρεί ότι

$$
\left.
\frac{\partial u}{\partial t}
\right|_{i,j+\frac{1}{2}}
=
\left.
\frac{\partial^2u}{\partial x^2}
\right|_{i,j+\frac{1}{2}}
$$

Η χρονική παράγωγος προσεγγίζεται μεταξύ των χρονικών στιγμών $t_j$ και $t_{j+1}$

$$
\left.
\frac{\partial u}{\partial t}
\right|_{i,j+\frac{1}{2}}
\approx
\frac{u_{i,j+1}-u_{i,j}}{k}
$$


### Διακριτοποίηση της χωρικής παραγώγου

Δεν διαθέτουμε άμεσα τις τιμές της λύσης στη χρονική στιγμή $t_{j+\frac{1}{2}}$.

Για τον λόγο αυτό, προσεγγίζουμε τη δεύτερη χωρική παράγωγο στο μέσο του χρονικού βήματος από τον μέσο όρο των αντίστοιχων παραγώγων στις χρονικές στιγμές $t_j$ και $t_{j+1}$

$$
\left.
\frac{\partial^2u}{\partial x^2}
\right|_{i,j+\frac{1}{2}}
\approx
\frac{1}{2}
\left(
\left.
\frac{\partial^2u}{\partial x^2}
\right|_{i,j}
+
\left.
\frac{\partial^2u}{\partial x^2}
\right|_{i,j+1}
\right)
$$

Στην τρέχουσα χρονική στιγμή $t_j$

$$
\left.
\frac{\partial^2u}{\partial x^2}
\right|_{i,j}
\approx
\frac{u_{i-1,j}-2u_{i,j}+u_{i+1,j}}{h^2}
$$

Στην επόμενη χρονική στιγμή $t_{j+1}$

$$
\left.
\frac{\partial^2u}{\partial x^2}
\right|_{i,j+1}
\approx
\frac{u_{i-1,j+1}-2u_{i,j+1}+u_{i+1,j+1}}{h^2}
$$

### Αντικατάσταση στην εξίσωση της θερμότητας

Αντικαθιστώντας τις αριθμητικές προσεγγίσεις προκύπτει

$$
\frac{u_{i,j+1}-u_{i,j}}{k}
=
\frac{1}{2}
\left(
\frac{u_{i-1,j+1}-2u_{i,j+1}+u_{i+1,j+1}}{h^2}
+
\frac{u_{i-1,j}-2u_{i,j}+u_{i+1,j}}{h^2}
\right)
$$

Ορίζουμε

$$
r=\frac{k}{h^2}
$$

Πολλαπλασιάζοντας με $2k$

$$
2u_{i,j+1}-2u_{i,j}
=
r
\left(
u_{i-1,j+1}
-2u_{i,j+1}
+u_{i+1,j+1}
\right)
+
r
\left(
u_{i-1,j}
-2u_{i,j}
+u_{i+1,j}
\right)
$$

Μεταφέρουμε τους άγνωστους όρους της επόμενης χρονικής στιγμής στο αριστερό μέλος και τους γνωστούς όρους της τρέχουσας χρονικής στιγμής στο δεξί.

$$
-r\,u_{i-1,j+1}
+
(2+2r)u_{i,j+1}
-
r\,u_{i+1,j+1}
=
r\,u_{i-1,j}
+
(2-2r)u_{i,j}
+
r\,u_{i+1,j}
$$

Η θερμοκρασία κάθε σημείου στην επόμενη χρονική στιγμή εξαρτάται από:

- τις άγνωστες θερμοκρασίες των γειτονικών σημείων στην επόμενη χρονική στιγμή

- τις γνωστές θερμοκρασίες των αντίστοιχων σημείων στην τρέχουσα χρονική στιγμή

Επομένως, η μέθοδος Crank – Nicolson είναι και αυτή implicit και απαιτεί την επίλυση συστήματος γραμμικών εξισώσεων.


#### Σύγκριση των 3 μεθόδων αριθμητικής επίλυσης της εξίσωσης θερμότητας

| Μέθοδος | Χρονική προσέγγιση της χωρικής παραγώγου | Ευστάθεια |
|:--|:--|:--|
| Explicit FTCS | $t_j$ | Υπό συνθήκη ευσταθής ($r\leq 1/2$) |
| Implicit BTCS | $t_{j+1}$ | Ευσταθής |
| Implicit Crank – Nicolson | Μέσος όρος στα $t_j$ και $t_{j+1}$ | Ευσταθής |

Οι μέθοδοι FTCS και BTCS είναι πρώτης τάξης ακρίβειας ως προς τον χρόνο.

Η μέθοδος Crank – Nicolson είναι δεύτερης τάξης ακρίβειας ως προς τον χρόνο, καθώς χρησιμοποιεί συμμετρική προσέγγιση γύρω από το μέσο του χρονικού βήματος.

Και οι τρεις μέθοδοι είναι δεύτερης τάξης ακρίβειας ως προς τον χώρο.
